### Decomposition of Diatomics: Table 1

```python
#Be (4e, 5o)
xyz = """
Be 0.00 0.00 0.00
"""

system = forte2.System(xyz=xyz, basis_set="cc-pVTZ", auxiliary_basis_set="def2-universal-JKFIT")

rhf = forte2.RHF(charge=0)(system)
rhf.run()

avas=forte2.AVAS(selection_method="separate",
                num_active_docc=2,
                num_active_uocc=3,
                subspace=["Be(1s)","Be(2s)","Be(2p)"])(rhf)

ci=forte2.CISolver(forte2.State(system=system, multiplicity=1,ms=0.0))
mc=forte2.MCOptimizer(ci, 
final_orbitals='natural'
)(avas)
mc.run()


frag_orbs = ci.mo_space.active_indices

ints = forte2.jkbuilder.RestrictedMOIntegrals(
    system=ci.system,
    C=ci.mos.C[0],
    orbitals=frag_orbs,
    core_orbitals=ci.core_indices,
)

e = ints.E #core energy
e1 = 0
e2 = 0
e3 = 0
e4 = 0


for i in frag_orbs:
    e1 += onefrag_correlation_energy_enumerated(ci, [i], core_orbitals=ci.core_indices)
    for j in frag_orbs:
        if j<i:
            e2 += twofrag_correlation_energy_enumerated(ci, [i], [j], core_orbitals=ci.core_indices)
            for k in frag_orbs:
                if k<j:
                    e3 += threefrag_correlation_energy_enumerated(ci, [i], [j], [k], core_orbitals=ci.core_indices)
                    for l in frag_orbs:
                        if l<k:
                            e4 += fourfrag_correlation_energy_enumerated(ci, [i], [j], [k], [l],core_orbitals=ci.core_indices)
print("========================")
print(f"E_core: {e}")
print(f"E1 Energy: {e1}")
print(f"E2 Energy: {e2}")
print(f"E3 Energy: {e3}")
print(f"E4 Energy: {e4}")
print(f"Total Decomposition Energy: {e1+e2+e3+e4+e}")
print(f"CI Energy: {ci.E[0]}")
```

### Breakdown of N2 Interactions

```python
#N2 (10e, 8o)
xyz = """
N 0.00 0.00 0.00
N 0.00 0.00 1.0977
"""

system = forte2.System(xyz=xyz, basis_set="cc-pVTZ", auxiliary_basis_set="def2-universal-JKFIT")

rhf = forte2.RHF(charge=0)(system)
rhf.run()

avas=forte2.AVAS(selection_method="separate",
                num_active_docc=5,
                num_active_uocc=3,
                subspace=["N(2s)","N(2p)"])(rhf)

ci=forte2.CISolver(forte2.State(system=system, multiplicity=1,ms=0.0))
mc=forte2.MCOptimizer(ci, 
final_orbitals='natural'
)(avas)
mc.run()
```

```python
frag_orbs = ci.mo_space.active_indices

ints = forte2.jkbuilder.RestrictedMOIntegrals(
    system=ci.system,
    C=ci.mos.C[0],
    orbitals=frag_orbs,
    core_orbitals=ci.core_indices,
)

e = ints.E #core energy
e1 = 0
e2 = 0
e3 = 0
e4 = 0


for i in frag_orbs:
    energy = onefrag_correlation_energy_enumerated(ci, [i], core_orbitals=ci.core_indices)
    e1 += energy
    if abs(energy) > 1e-8:
        print(f"One-Fragment {i} Energy: {energy}")


for i in frag_orbs:
    for j in frag_orbs:
        if j<i:
            energy = twofrag_correlation_energy_enumerated(ci, [i], [j], core_orbitals=ci.core_indices)
            e2 += energy
            if abs(energy) > 1e-8:
                print(f"Two-Fragment {i}, {j} Energy: {energy}")

for i in frag_orbs:
    for j in frag_orbs:
        if j<i:
            for k in frag_orbs:
                if k<j:
                    energy = threefrag_correlation_energy_enumerated(ci, [i], [j], [k], core_orbitals=ci.core_indices)
                    e3 += energy
                    if abs(energy) > 1e-8:
                        print(f"Three-Fragment {i}, {j}, {k} Energy: {energy}")

for i in frag_orbs:
    for j in frag_orbs:
        if j<i:
            for k in frag_orbs:
                if k<j:
                    for l in frag_orbs:
                        if l<k:
                            energy = fourfrag_correlation_energy_enumerated(ci, [i], [j], [k], [l],core_orbitals=ci.core_indices)
                            e4 += energy
                            if abs(energy) > 1e-8:
                                print(f"Four-Fragment {i}, {j}, {k}, {l} Energy: {energy}")


print("========================")
print(f"E_core: {e}")
print(f"E1 Energy: {e1}")
print(f"E2 Energy: {e2}")
print(f"E3 Energy: {e3}")
print(f"E4 Energy: {e4}")
print(f"Total Decomposition Energy: {e1+e2+e3+e4+e}")
print(f"CI Energy: {ci.E[0]}")
```
